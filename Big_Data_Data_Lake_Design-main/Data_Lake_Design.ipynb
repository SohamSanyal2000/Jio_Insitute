

**1. Find revenue generated by different categories for the month of 11/2020.**

```
from pyspark.sql import SparkSession
from pyspark.sql.functions import round

# Assuming that spark is an instance of SparkSession
df = spark.table("orderstranscpt")

result = (df
    .filter(df.month == '2020-11-01')
    .groupBy(df.category)
    .agg(round(F.sum(df.total), 4).alias('Revenue'))
)

result.show()
```



**2) Which top 5 categories have a maximum number of refunds in the year 2020?**

```
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()

df = spark.table("orderstranscpt")

result = (df
          .filter((df.year == 2020) & (df.status == 'order_refunded'))
          .groupBy("category")
          .count()
          .withColumnRenamed("count", "refund_count")
          .orderBy("refund_count", ascending=False)
          .limit(5))

result.show()
```



**3) Find a total number of orders by each category for each month and year in the dataset?**



**4. Segment customers by age: 0-20 as young, 20-35 adults, 35-55 middle-ages and >55 Old. Find the total spend (in percentage of total spend of categories) by** 
**customers by different age segments by different categories.**

**E.g., Categories Total Spend Young Adults Middle-Age Old**

**E.g., In appliances category total spend is X amount of which a,b,c,d are percentages contributed from Young, Adults, Middle-Age and Old respectively. (a+b+c+d=100)**

```
from pyspark.sql import SparkSession
from pyspark.sql.functions import when, sum, col

# Assuming that SparkSession is already initialized
spark = SparkSession.builder.getOrCreate()

# Load the data into DataFrame
df = spark.read.table("orderstranscpt")

# Create age segments
df = df.withColumn("age_segment", 
                   when(col("age") <= 20, "Young")
                   .when(col("age") <= 35, "Adults")
                   .when(col("age") <= 55, "Middle-Age")
                   .otherwise("Old"))

# Calculate total spend by category
total_spend_df = df.groupBy("category").agg(sum("total").alias("total_spend"))

# Join with original DataFrame and calculate percentage by category
result_df = df.join(total_spend_df, on="category", how="inner")
result_df = result_df.withColumn("per_by_category", (col("total") / col("total_spend")) * 100)

# Group by age_segment and category and calculate sum
result_df = result_df.groupBy("age_segment", "category").agg(sum("per_by_category").alias("per_by_category"))

# Order by category
result_df = result_df.orderBy("category")

# Show the result
result_df.show()
```

**5)Spend by gender across different categories in terms of percentages.**

```
from pyspark.sql import functions as F

# Load the data into a DataFrame
df = spark.table("orderstranscpt")

# Create the CategorySpend DataFrame
CategorySpend = df.groupBy("category", "gender").agg(F.sum("total").alias("total_spend"))

# Create the TotalSpendByCategory DataFrame
TotalSpendByCategory = df.groupBy("category").agg(F.sum("total").alias("category_total"))

# Join the two DataFrames and calculate the spend_percentage
result = CategorySpend.join(TotalSpendByCategory, "category") \
    .withColumn("spend_percentage", (F.col("total_spend") / F.col("category_total")) * 100) \
    .orderBy("category", "gender")

# Show the result
result.show()
```

**6. Find top 5 customers for each month. The results should be as follows: E.g., Year, Month, Custid, Name, country, Gender, Total Spend, Rank**

```
from pyspark.sql import SparkSession
from pyspark.sql.functions import sum as _sum, rank
from pyspark.sql.window import Window

spark = SparkSession.builder.getOrCreate()

# Load your data into a DataFrame
df = spark.read.table("orderstranscpt")

# Create the MonthlyCustomerSpend DataFrame
MonthlyCustomerSpend = df.groupBy("year", "month", "cust_id", "full_name", "County", "Gender")\
    .agg(_sum("value").alias("TotalSpend"))

# Create the RankedCustomers DataFrame
windowSpec = Window.partitionBy("year", "month").orderBy(MonthlyCustomerSpend["TotalSpend"].desc())
RankedCustomers = MonthlyCustomerSpend.withColumn("Rank", rank().over(windowSpec))

# Filter for the top 5 customers each month and sort
result = RankedCustomers.filter(RankedCustomers["Rank"] <= 5)\
    .orderBy("year", "month", "Rank")

result.show()
```


**7. Calculate the RFM values for each customer (by customer id).
RFM Represents R (Recency) F (Frequency) M (Monetary Value) (Click on the link to read more about RFM)
For Recency calculation, use 31/10/2021 as last date. So, the recency for any customer should be how many weeks before he or she has made the last purchase from the date of 01/10/2021.**

```
from pyspark.sql import SparkSession
from pyspark.sql.functions import datediff, max, countDistinct, sum

spark = SparkSession.builder.getOrCreate()

# Assuming 'orderstranscpt' is a DataFrame
orderstranscpt = spark.table("orderstranscpt")

CustomerRFM = (orderstranscpt
               .groupBy("cust_id")
               .agg(datediff(max("order_date"), lit("2021-10-01")).alias("Recency"),
                    countDistinct("order_id").alias("Frequency"),
                    sum("value").alias("Monetary")))

CustomerRFM.show()
```

#### 8) Top 10 customers based on frequency and monetary value. Sort them based on first frequency and then monetary value

```
from pyspark.sql import SparkSession
from pyspark.sql.functions import sum, countDistinct

spark = SparkSession.builder.getOrCreate()

orderstranscpt = spark.table("orderstranscpt")

CustomerFrequencyMonetary = (orderstranscpt
    .groupBy("cust_id")
    .agg(
        countDistinct("order_id").alias("Frequency"),
        sum("value").alias("Monetary")
    )
)

result = (CustomerFrequencyMonetary
    .orderBy(CustomerFrequencyMonetary["Frequency"].desc(), CustomerFrequencyMonetary["Monetary"].desc())
    .limit(10)
)

result.show()
```
